```python
import matplotlib.pyplot as plt

land_cover = {
    "Bare rock / scree":       0.90,
    "Roads / buildings":       0.85,
    "Sparse grassland":        0.55,
    "Terraced agriculture":    0.35,
    "Shrubland":               0.30,
    "Dense broadleaf forest":  0.10,
    "Water bodies":            0.05,
}

fig, ax = plt.subplots(figsize=(10, 4))
names = list(land_cover.keys())
values = list(land_cover.values())
colors = ['green' if v > 0.5 else 'orange' if v > 0.25 else 'red' for v in values]

ax.barh(names, values, color=colors, alpha=0.8, edgecolor='black')
ax.axvline(x=0.3, color='red', ls='--', lw=2, label='Usability threshold')
ax.set_xlabel("Expected Coherence (0-1)")
ax.set_title("InSAR Coherence by Land Cover (Nepal)")
ax.legend()
ax.set_xlim(0, 1)
plt.tight_layout()
plt.show()
print("GREEN = works. ORANGE = marginal. RED = unusable.")

# If study area is mostly forest (red), NEED time-series stacking, If mostly bare rock (green), single-pair DInSAR gives clean results.
```
