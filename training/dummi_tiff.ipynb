```python
import numpy as np
import rasterio
from rasterio.transform import from_bounds
```

```python
west, south, east, north = 318236.9, 3051179.8, 319553.0, 3052232.7 # Meters (Bbox: West, South, East, North)

width, height = 100, 100
data = np.linspace(-10, 10, width*height).reshape(height,width)
# displacement signal in mm
transform = from_bounds(west, south, east, north, width, height)

with rasterio.open(
    "../data/test_displacement.tif", "w", driver = "GTiff", height=height, width=width, count=1, dtype="float32", crs="EPSG:32645", transform=transform,) as dst:
    dst.write(data,1)

# count = no of bands/layers (1 = single measurement)

print("✅ GeoTIFF written: data/test_displacement.tif")

```
